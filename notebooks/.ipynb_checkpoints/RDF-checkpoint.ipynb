```python
# TGraph records carry attributes; convert to geometry only when needed.
def _as_topology(value):
    from topologicpy.Topology import Topology
    from topologicpy.Vertex import Vertex
    if isinstance(value, (list, tuple)):
        return [_as_topology(item) for item in value]
    if not isinstance(value, dict) or "index" not in value or "dictionary" not in value:
        return value
    representation = value.get("representation")
    attributes = value["dictionary"]
    if representation is None and "src" not in value:
        representation = Vertex.ByCoordinates(
            attributes.get("x", 0), attributes.get("y", 0), attributes.get("z", 0))
    if representation is None:
        raise ValueError("This graph record has no geometry representation")
    return Topology.SetDictionary(representation, attributes)


def _record_dictionary(value):
    from topologicpy.Topology import Topology
    if isinstance(value, dict) and "dictionary" in value:
        return value["dictionary"]
    return Topology.Dictionary(value)
```

```python
# Make sure you have pip installed rdflib

from topologicpy.Vertex import Vertex
from topologicpy.Edge import Edge
from topologicpy.Wire import Wire
from topologicpy.Face import Face
from topologicpy.Shell import Shell
from topologicpy.Cell import Cell
from topologicpy.CellComplex import CellComplex
from topologicpy.Cluster import Cluster
from topologicpy.Topology import Topology
from topologicpy.Aperture import Aperture
from topologicpy.TGraph import TGraph
from topologicpy.Dictionary import Dictionary
from topologicpy.Helper import Helper
from topologicpy.Plotly import Plotly

import json
from rdflib import TGraph as RDFGraph
from rdflib import URIRef, Literal, Namespace
from rdflib.namespace import RDF, RDFS


def Graph_RDF(graph,
                    bidirectional=False,
                    includeAttributes=False,
                    includeLabel=False,
                    typeKey="type",
                    labelKey="label",
                    sourceKey="source",
                    targetKey="target",
                    floorLevels =[],
                    format="turtle"):
    #     turtle, ttl or turtle2 : Turtle, turtle2 is just turtle with more spacing & linebreaks
    #     xml or pretty-xml : RDF/XML, Was the default format, rdflib < 6.0.0
    #     json-ld : JSON-LD , There are further options for compact syntax and other JSON-LD variants
    #     ntriples, nt or nt11 : N-Triples , nt11 is exactly like nt, only utf8 encoded
    #     n3 : Notation-3 , N3 is a superset of Turtle that also caters for rules and a few other things
    #     trig : Trig , Turtle-like format for RDF triples + context (RDF quads) and thus multiple graphs
    #     trix : Trix , RDF/XML-like format for RDF quads
    #     nquads : N-Quads , N-Triples-like format for RDF quads

    json_data = TGraph.JSONData(graph)
    # Create an empty RDF graph
    rdf_graph = RDFGraph()
    
    # Define namespaces
    rdf = Namespace("http://www.w3.org/1999/02/22-rdf-syntax-ns#")
    rdfs = Namespace("http://www.w3.org/2000/01/rdf-schema#")
    bot = Namespace("https://w3id.org/bot#")
    
    # Define a custom prefix mapping
    rdf_graph.namespace_manager.bind("bot", bot)
    
    # Add site
    site_uri = URIRef("Site_0001")
    rdf_graph.add((site_uri, rdf.type, bot.Site))
    if includeLabel:
        rdf_graph.add((site_uri, RDFS.label, Literal("Site_0001")))
    
    # Add building
    building_uri = URIRef("Building_0001")
    rdf_graph.add((building_uri, rdf.type, bot.Building))
    if includeLabel:
        rdf_graph.add((building_uri, RDFS.label, Literal("Building_0001")))
    
    # Add stories
    storey_uris = []
    n = max(len(str(len(floorLevels))),4)
    for i, floor_level in enumerate(floorLevels):
        storey_uri = URIRef("Storey_"+str(i+1).zfill(n))
        rdf_graph.add((storey_uri, rdf.type, bot.Storey))
        if includeLabel:
            rdf_graph.add((storey_uri, RDFS.label, Literal("Storey_"+str(i+1).zfill(n))))
        storey_uris.append(storey_uri)

    # Add triples to relate building to site and stories to building
    rdf_graph.add((site_uri, bot.hasBuilding, building_uri))
    if bidirectional:
        rdf_graph.add((building_uri, bot.isPartOf, site_uri)) # might not be needed

    for storey_uri in storey_uris:
        rdf_graph.add((building_uri, bot.hasStorey, storey_uri))
        if bidirectional:
            rdf_graph.add((storey_uri, bot.isPartOf, building_uri)) # might not be needed
    
    # Add vertices as RDF resources
    for node, attributes in json_data['nodes'].items():
        node_uri = URIRef(node)
        if 'space' in attributes[typeKey].lower():
            rdf_graph.add((node_uri, rdf.type, bot.Space))
            # Find the storey it is on
            z = attributes['z']
            level = Helper.Position(z, floorLevels)
            if level > len(storey_uris):
                level = len(storey_uris)
            storey_uri = storey_uris[level-1]
            rdf_graph.add((storey_uri, bot.hasSpace, node_uri))
            if bidirectional:
                rdf_graph.add((node_uri, bot.isPartOf, storey_uri)) # might not be needed
        elif 'window' in attributes[typeKey].lower():
            rdf_graph.add((node_uri, rdf.type, bot.Window))
        elif 'door' in attributes[typeKey].lower():
            rdf_graph.add((node_uri, rdf.type, bot.Door))
        elif 'wall' in attributes[typeKey].lower():
            rdf_graph.add((node_uri, rdf.type, bot.Wall))
        elif 'slab' in attributes[typeKey].lower():
            rdf_graph.add((node_uri, rdf.type, bot.Slab))
        else:
            rdf_graph.add((node_uri, rdf.type, bot.BuildingElement))
        
        if includeAttributes:
            for key, value in attributes.items():
                if key == labelKey:
                    if includeLabel:
                        rdf_graph.add((node_uri, RDFS.label, Literal(value)))
                elif key != typeKey:
                    rdf_graph.add((node_uri, bot[key], Literal(value)))
        if includeLabel:
            for key, value in attributes.items():
                if key == labelKey:
                    rdf_graph.add((node_uri, RDFS.label, Literal(value)))
    
    # Add edges as RDF triples
    for edge, attributes in json_data['edges'].items():
        source = attributes[sourceKey]
        target = attributes[targetKey]
        source_uri = URIRef(source)
        target_uri = URIRef(target)
        if "space" in json_data['nodes'][source][typeKey].lower() and "space" in json_data['nodes'][target][typeKey].lower():
            rdf_graph.add((source_uri, bot.adjacentTo, target_uri))
            if bidirectional:
                rdf_graph.add((target_uri, bot.adjacentTo, source_uri))
        elif "space" in json_data['nodes'][source][typeKey].lower() and "wall" in json_data['nodes'][target][typeKey].lower():
            rdf_graph.add((target_uri, bot.interfaceOf, source_uri))
        elif "space" in json_data['nodes'][source][typeKey].lower() and "slab" in json_data['nodes'][target][typeKey].lower():
            rdf_graph.add((target_uri, bot.interfaceOf, source_uri))
        else:
            rdf_graph.add((source_uri, bot.connectsTo, target_uri))
        #for key, value in attributes.items():
            #rdf_graph.add((source_uri, bot[key], Literal(value)))
    # Return serialized  RDF graph
    rdf_serialized = rdf_graph.serialize(format=format)
    return rdf_serialized


```

```python
def assignTypesAndLabels(cellComplex, dictionary, typeKey="type", labelKey="label"):
    decomp = CellComplex.Decompose(cellComplex)
    keys = decomp.keys()
    for key in keys:
        typeValue = dictionary[key]
        elements = decomp[key]
        n = max(len(str(len(elements))),4)
        for i, element in enumerate(elements):
            d = _record_dictionary(element)
            labelValue = Dictionary.ValueAtKey(d, labelKey)
            if labelValue == None:
                labelValue = typeValue+"_"+str(i+1).zfill(n)
            d = Dictionary.ByKeysValues([typeKey, labelKey], [typeValue, labelValue])
            element = Topology.SetDictionary(element, d)
```

## Build the House

```python
selectors = []

lr_origin = Vertex.ByCoordinates(0, 0, 0)
living_room = Cell.Prism(origin=lr_origin, width=3, length=5, height=3, placement="lowerleft")
cen = Topology.Centroid(_as_topology(living_room))
d = Dictionary.ByKeyValue("label", "Living_Room_001")
cen = Topology.SetDictionary(cen, d)
selectors.append(cen)

kt_origin = Vertex.ByCoordinates(0,5,0)
kitchen = Cell.Prism(origin=kt_origin, width=3, length=4, height=3, placement="lowerleft")
cen = Topology.Centroid(_as_topology(kitchen))
d = Dictionary.ByKeyValue("label", "Kitchen_001")
cen = Topology.SetDictionary(cen, d)
selectors.append(cen)

cr_origin = Vertex.ByCoordinates(3,2,0)
corridor = Cell.Prism(origin=cr_origin, width=2, length=6, height=3, placement="lowerleft")
cen = Topology.Centroid(_as_topology(corridor))
d = Dictionary.ByKeyValue("label", "Corridor_001")
cen = Topology.SetDictionary(cen, d)
selectors.append(cen)

br_origin = Vertex.ByCoordinates(5,0,0)
bedroom = Cell.Prism(origin=br_origin, width=4, length=4, height=3, placement="lowerleft")
cen = Topology.Centroid(_as_topology(bedroom))
d = Dictionary.ByKeyValue("label", "Bedroom_001")
cen = Topology.SetDictionary(cen, d)
selectors.append(cen)

bt_origin = Vertex.ByCoordinates(5,4,0)
bathroom = Cell.Prism(origin=bt_origin, width=3, length=2, height=3, placement="lowerleft")
cen = Topology.Centroid(_as_topology(bathroom))
d = Dictionary.ByKeyValue("label", "Bathroom_001")
cen = Topology.SetDictionary(cen, d)
selectors.append(cen)

ho_origin = Vertex.ByCoordinates(5,6,0)
home_office = Cell.Prism(origin=ho_origin, width=4, length=3, height=3, placement="lowerleft")
cen = Topology.Centroid(_as_topology(home_office))
d = Dictionary.ByKeyValue("label", "Home_Office_001")
cen = Topology.SetDictionary(cen, d)
selectors.append(cen)

sw_origin = Vertex.ByCoordinates(3.2,8,0)
stairwell = Cell.Prism(origin=sw_origin, width=1.6, length=3.2, height=6, wSides=2, placement="lowerleft")
cen = Topology.Centroid(_as_topology(stairwell))
d = Dictionary.ByKeyValue("label", "Stairwell_001")
cen = Topology.SetDictionary(cen, d)
selectors.append(cen)

gd_flr_rooms = [living_room, kitchen, corridor, bedroom, bathroom, home_office]

lr_origin = Vertex.ByCoordinates(0,0,3)
living_room = Cell.Prism(origin=lr_origin, width=3, length=5, height=3, placement="lowerleft")
cen = Topology.Centroid(_as_topology(living_room))
d = Dictionary.ByKeyValue("label", "Living_Room_002")
cen = Topology.SetDictionary(cen, d)
selectors.append(cen)

kt_origin = Vertex.ByCoordinates(0,5,3)
kitchen = Cell.Prism(origin=kt_origin, width=3, length=4, height=3, placement="lowerleft")
cen = Topology.Centroid(_as_topology(kitchen))
d = Dictionary.ByKeyValue("label", "Kitchen_002")
cen = Topology.SetDictionary(cen, d)
selectors.append(cen)

cr_origin = Vertex.ByCoordinates(3,2,3)
corridor = Cell.Prism(origin=cr_origin, width=2, length=6, height=3, placement="lowerleft")
cen = Topology.Centroid(_as_topology(corridor))
d = Dictionary.ByKeyValue("label", "Corridor_002")
cen = Topology.SetDictionary(cen, d)
selectors.append(cen)

br_origin = Vertex.ByCoordinates(5,0,3)
bedroom = Cell.Prism(origin=br_origin, width=4, length=4, height=3, placement="lowerleft")
cen = Topology.Centroid(_as_topology(bedroom))
d = Dictionary.ByKeyValue("label", "Bedroom_002")
cen = Topology.SetDictionary(cen, d)
selectors.append(cen)

bt_origin = Vertex.ByCoordinates(5,4,3)
bathroom = Cell.Prism(origin=bt_origin, width=3, length=2, height=3, placement="lowerleft")
cen = Topology.Centroid(_as_topology(bathroom))
d = Dictionary.ByKeyValue("label", "Bathroom_002")
cen = Topology.SetDictionary(cen, d)
selectors.append(cen)

ho_origin = Vertex.ByCoordinates(5,6,3)
home_office = Cell.Prism(origin=ho_origin, width=4, length=3, height=3, placement="lowerleft")
cen = Topology.Centroid(_as_topology(home_office))
d = Dictionary.ByKeyValue("label", "Home_Office_002")
cen = Topology.SetDictionary(cen, d)
selectors.append(cen)


fst_flr_rooms = [living_room, kitchen, corridor, bedroom, bathroom, home_office]

house = CellComplex.ByCells(gd_flr_rooms+fst_flr_rooms+[stairwell])

# Build the doors
orig = Vertex.ByCoordinates(0.2,5,0.05)
d2 = Face.Rectangle(origin=orig, width=2.6, length=2.10, placement="lowerleft")
d2 = Topology.Rotate(_as_topology(d2), origin=orig, axis=[1, 0, 0], angle=90)

orig = Vertex.ByCoordinates(3,3.4,0.05)
d3 = Face.Rectangle(origin=orig, width=2.10, length=1, placement="lowerleft")
d3 = Topology.Rotate(_as_topology(d3), origin=orig, axis=[0, 1, 0], angle=-90)

orig = Vertex.ByCoordinates(3,5.2,0.05)
d4 = Face.Rectangle(origin=orig, width=2.10, length=1, placement="lowerleft")
d4 = Topology.Rotate(_as_topology(d4), origin=orig, axis=[0, 1, 0], angle=-90)

orig = Vertex.ByCoordinates(5,2.8,0.05)
d5 = Face.Rectangle(origin=orig, width=2.10, length=1, placement="lowerleft")
d5 = Topology.Rotate(_as_topology(d5), origin=orig, axis=[0, 1, 0], angle=-90)

orig = Vertex.ByCoordinates(5,4.2,0.05)
d6 = Face.Rectangle(origin=orig, width=2.10, length=1, placement="lowerleft")
d6 = Topology.Rotate(_as_topology(d6), origin=orig, axis=[0, 1, 0], angle=-90)

orig = Vertex.ByCoordinates(5,6.2,0.05)
d7 = Face.Rectangle(origin=orig, width=2.10, length=1, placement="lowerleft")
d7 = Topology.Rotate(_as_topology(d7), origin=orig, axis=[0, 1, 0], angle=-90)

orig = Vertex.ByCoordinates(3.3,2,0.05)
d8 = Face.Rectangle(origin=orig, width=1.4, length=2.10, placement="lowerleft")
d8 = Topology.Rotate(_as_topology(d8), origin=orig, axis=[1, 0, 0], angle=90)
door_centroid = Topology.Centroid(_as_topology(d8))
gd_flr_drs = [d2,d3,d4,d5,d6,d7]
fst_flr_dr = [Topology.Translate(_as_topology(door), 0, 0, 3) for door in gd_flr_drs]
drs = gd_flr_drs + fst_flr_dr + [d8]

# Set the dictionaries for the doors
doors = []
for door in drs:
    #d = Dictionary.ByKeyValue("type", "Door")
    #door = Topology.SetDictionary(door, d)
    doors.append(door)
door_cluster = Cluster.ByTopologies(_as_topology(doors))

decomp = CellComplex.Decompose(house)
outer_walls = decomp["externalVerticalFaces"]
windows = []
for outer_wall in outer_walls:
    window = Topology.Scale(_as_topology(outer_wall), origin=Topology.Centroid(_as_topology(outer_wall)), x=0.3, y=0.3, z=0.3)
    window = Topology.Translate(_as_topology(window), x=0, y=0, z=-0.4)
    #d = Dictionary.ByKeyValue("type", "Window")
    #window = Topology.SetDictionary(window, d)
    windows.append(window)
windows = [w for w in windows if Vertex.Distance(_as_topology(Topology.Centroid(_as_topology(w))), _as_topology(door_centroid)) > 0.2]
windows = [w for w in windows if Face.Area(w) > 0.5]

window_cluster = Cluster.ByTopologies(_as_topology(windows))
#Assign Room Dictionaries to the Cells in the CellComplex
house = Topology.TransferDictionariesBySelectors(house, selectors, tranCells=True)

# Add the doors and windows to the house
house = Topology.AddApertures(house, doors+windows, exclusive=True, subTopologyType="face", tolerance=0.001)

d = {
    "cells": "Space",
    "externalVerticalFaces": "External_Vertical_Wall",
    "internalVerticalFaces": "Internal_Vertical_Wall",
    "topHorizontalFaces": "Roof_Slab",
    "bottomHorizontalFaces": "Ground_Slab",
    "internalHorizontalFaces": "Internal_Floor_Slab",
    "externalInclinedFaces": "External_Inclined_Wall",
    "internalInclinedFaces": "Internal_Inclined_Wall",
    "externalVerticalApertures": "External_Vertical_Window",
    "internalVerticalApertures": "Internal_Vertical_Door",
    "topHorizontalApertures": "Skylight_Roof_Window",
    "bottomHorizontalApertures": "Floor_Window",
    "internalHorizontalApertures": "Floor_Window",
    "externalInclinedApertures": "External_Inclined_Window",
    "internalInclinedApertures": "Internal_Inclined_Door",
}
#Assign the appropriate element types
_ = assignTypesAndLabels(house, d, typeKey="type", labelKey="label")
# Visualize the house
Topology.Show(house, doors, windows, faceOpacity=0.5)
```

## Visualize the Graph of the House

```python
house_graph = TGraph.ByTopology(house,
                     direct=False,
                     directApertures=True,
                     toExteriorTopologies=True,
                     viaSharedTopologies=True,
                     toExteriorApertures=True,
                     viaSharedApertures=True)
data01 = Plotly.DataByTopology(house, faceOpacity=0.1)
data02 = Plotly.DataByTopology(door_cluster, faceColor="brown", faceOpacity=0.2)
data03 = Plotly.DataByTopology(window_cluster, faceColor="lightblue", faceOpacity=0.8)
data04= Plotly.DataByTopology(TGraph.Topology(house_graph), vertexSize=5, vertexColor="red", edgeWidth=2)
figure = Plotly.FigureByData(data01+data02+data03+data04)
Plotly.Show(figure)
```

## Compute the List of Floor Levels

```python
floor_levels = []
slabs = decomp['bottomHorizontalFaces'] + decomp['internalHorizontalFaces']
for slab in slabs:
    floor_levels.append(Vertex.Z(_as_topology(Topology.Centroid(_as_topology(slab)))))
floor_levels = list(set(floor_levels))
floor_levels.sort()
print("Floor Levels:", floor_levels)
```

## Compute the RDF String

```python
rdf_string = Graph_RDF(graph=house_graph,
                       bidirectional=True,
                       includeAttributes=True,
                       includeLabel=True,
                       typeKey="type",
                       labelKey="label",
                       sourceKey="source",
                       targetKey="target",
                       floorLevels = floor_levels,
                       format="turtle")
print(rdf_string)
```
