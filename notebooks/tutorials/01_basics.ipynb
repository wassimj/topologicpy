Create a point and a line, then inspect their types. The first cell explicitly selects pythonocc: it fails if the requested backend is unavailable, rather than silently using another kernel.

```python
import os
os.environ["TOPOLOGICPY_CORE_BACKEND"] = "pythonocc"

import topologicpy
from topologicpy.Core import Core

assert topologicpy.__version__ == "0.9.86", "Use TopologicPy 0.9.86 for these tutorials."
assert type(Core.Backend()).__name__ == "PythonOCCBackend", "Restart with the pythonocc backend."
print("TopologicPy", topologicpy.__version__, "—", type(Core.Backend()).__name__)
```

## Make two vertices and an edge

A vertex represents a point. An edge connects two distinct vertices. Use the TopologicPy classes rather than importing a kernel directly.


```python
from topologicpy.Vertex import Vertex
from topologicpy.Edge import Edge
from topologicpy.Topology import Topology

start = Vertex.ByCoordinates(0, 0, 0)
end = Vertex.ByCoordinates(3, 4, 0)
edge = Edge.ByVertices([start, end])

print("Start:", Vertex.Coordinates(start))
print("Type:", Topology.TypeAsString(edge))
print("Length:", Edge.Length(edge))
assert Edge.Length(edge) == 5
```

## Tolerances and common errors

A tolerance is a distance used when comparing geometry. Points closer than the tolerance may be treated as coincident. Avoid zero-length edges, use consistent units, and check returned objects before continuing. A version or backend assertion above usually means the wrong notebook kernel is selected.


```python
tolerance = 0.0001
nearby = Vertex.ByCoordinates(0.00001, 0, 0)
distance = Vertex.Distance(start, nearby)
print("Within tolerance:", distance < tolerance)
assert Topology.IsInstance(edge, "Edge")
```
