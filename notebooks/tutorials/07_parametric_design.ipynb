# 07. Parametric Design and Optimisation

Compare five room proportions with the same floor area. Use a small exhaustive search rather than a long optimisation run.

**Goal:** Complete one small workflow. Run the cells from top to bottom in a fresh kernel.

**Extra dependencies:** None beyond the base environment.

All examples use metres unless stated otherwise.


```python
import os
os.environ["TOPOLOGICPY_CORE_BACKEND"] = "pythonocc"

import topologicpy
from topologicpy.Core import Core

assert topologicpy.__version__ == "0.9.86", "Use TopologicPy 0.9.86 for these tutorials."
assert type(Core.Backend()).__name__ == "PythonOCCBackend", "Restart with the pythonocc backend."
print("TopologicPy", topologicpy.__version__, "—", type(Core.Backend()).__name__)
```

## Generate and evaluate alternatives

Keep floor area at 24 m² and height at 3 m. Our objective is to minimise total surface area, a geometric proxy rather than a complete building-performance measure.


```python
from topologicpy.Cell import Cell

floor_area = 24
height = 3
candidates = []
for width in [3, 4, 5, 6, 8]:
    length = floor_area / width
    room = Cell.Prism(width=width, length=length, height=height)
    surface_area = Cell.SurfaceArea(room)
    candidates.append([width, length, surface_area])
    print("Width:", width, "Length:", round(length, 2), "Surface area:", surface_area)
```

## Select the best candidate

A simple loop makes the selection rule visible. This finds the best of the five candidates, not a guaranteed global optimum over all possible dimensions.


```python
best = candidates[0]
for candidate in candidates:
    if candidate[2] < best[2]:
        best = candidate
print("Best width, length, surface area:", best)
assert best[0] == 5
assert abs(best[0] * best[1] - floor_area) < 0.0001
```
