# 04. Dictionaries and Semantic Data

Give a room a name and an occupancy, then read those values back.

**Goal:** Complete one small workflow. Run the cells from top to bottom in a fresh kernel.

**Extra dependencies:** None beyond the base environment.

All examples use metres unless stated otherwise.


```python
import os
os.environ["TOPOLOGICPY_CORE_BACKEND"] = "pythonocc"

import topologicpy
from topologicpy.Core import Core

assert topologicpy.__version__ == "0.9.86", "Use TopologicPy 0.9.86 for these tutorials."
assert type(Core.Backend()).__name__ == "PythonOCCBackend", "Restart with the pythonocc backend."
print("TopologicPy", topologicpy.__version__, "—", type(Core.Backend()).__name__)
```

## Attach meaning to geometry

A dictionary stores named values. Keep the object returned by `SetDictionary`.


```python
from topologicpy.Cell import Cell
from topologicpy.Dictionary import Dictionary
from topologicpy.Topology import Topology

room = Cell.Prism(width=4, length=3, height=3)
properties = Dictionary.ByKeysValues(["name", "occupancy"], ["Studio", 4])
room = Topology.SetDictionary(room, properties)
stored = Topology.Dictionary(room)
print("Name:", Dictionary.ValueAtKey(stored, "name"))
print("Occupancy:", Dictionary.ValueAtKey(stored, "occupancy"))
assert Dictionary.ValueAtKey(stored, "occupancy") == 4
```

## Preserve attributes during a transformation

Dictionary behaviour depends on the operation. Request transfer explicitly here and verify the result instead of assuming every operation preserves metadata.


```python
moved_room = Topology.Translate(room, x=5, transferDictionaries=True)
moved_properties = Topology.Dictionary(moved_room)
print("Moved room:", Dictionary.ValueAtKey(moved_properties, "name"))
assert Dictionary.ValueAtKey(moved_properties, "name") == "Studio"
```
