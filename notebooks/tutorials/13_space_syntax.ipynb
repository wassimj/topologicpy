# 13. Space Syntax Analysis

Compare integration and choice on a four-space access graph. This is an introduction to graph-based spatial interpretation.

**Goal:** Complete one small workflow. Run the cells from top to bottom in a fresh kernel.

**Extra dependencies:** None beyond the base environment.

All examples use metres unless stated otherwise.


```python
import os
os.environ["TOPOLOGICPY_CORE_BACKEND"] = "pythonocc"

import topologicpy
from topologicpy.Core import Core

assert topologicpy.__version__ == "0.9.86", "Use TopologicPy 0.9.86 for these tutorials."
assert type(Core.Backend()).__name__ == "PythonOCCBackend", "Restart with the pythonocc backend."
print("TopologicPy", topologicpy.__version__, "—", type(Core.Backend()).__name__)
```

## Create an access graph

Each vertex represents a space and each edge a direct connection. This is a justified access representation, not an axial or visibility graph.


```python
from topologicpy.Vertex import Vertex
from topologicpy.TGraph import TGraph

entrance = Vertex.ByCoordinates(0, 0, 0)
hall = Vertex.ByCoordinates(2, 0, 0)
studio = Vertex.ByCoordinates(4, 0, 0)
office = Vertex.ByCoordinates(2, 2, 0)
vertices = [entrance, hall, studio, office]
edges = [{"src": 0, "dst": 1},
         {"src": 1, "dst": 2},
         {"src": 1, "dst": 3}]
graph = TGraph.ByVerticesEdges(vertices, edges, directed=False)
assert isinstance(graph, TGraph)
assert len(TGraph.Vertices(graph)) == 4
assert len(TGraph.Edges(graph)) == 3
```

## Calculate two measures

`TGraph.Integration` defaults to closeness-based integration; `TGraph.Choice` uses vertex betweenness. Both use hop distances here. These introduce accessibility and through-movement, rather than every conventional space-syntax normalisation. The API also offers `method="hh"` for Hillier–Hanson integration, where some tiny graphs have undefined values.


```python
integration = TGraph.Integration(graph, method="closeness")
choice = TGraph.Choice(graph)
for index in range(len(vertices)):
    print("Space", index, "Integration:", integration[index], "Choice:", choice[index])
assert integration[1] > integration[0]
assert choice[1] > choice[0]
```
