# 06. Meshing and Discretisation

Triangulate a rectangle and inspect the indexed mesh of a box. This example avoids an external meshing engine.

**Goal:** Complete one small workflow. Run the cells from top to bottom in a fresh kernel.

**Extra dependencies:** None beyond the base environment.

All examples use metres unless stated otherwise.


```python
import os
os.environ["TOPOLOGICPY_CORE_BACKEND"] = "pythonocc"

import topologicpy
from topologicpy.Core import Core

assert topologicpy.__version__ == "0.9.86", "Use TopologicPy 0.9.86 for these tutorials."
assert type(Core.Backend()).__name__ == "PythonOCCBackend", "Restart with the pythonocc backend."
print("TopologicPy", topologicpy.__version__, "—", type(Core.Backend()).__name__)
```

## Split a face into triangles

Triangulation replaces a face with triangular faces. Their total area should equal the original area.


```python
from topologicpy.Face import Face
from topologicpy.Cell import Cell
from topologicpy.Topology import Topology

rectangle = Face.Rectangle(width=4, length=3)
triangles = Face.Triangulate(rectangle)
total_area = 0
for triangle in triangles:
    total_area = total_area + Face.Area(triangle)
print("Triangles:", len(triangles), "Total area:", total_area)
assert abs(total_area - Face.Area(rectangle)) < 0.0001
```

## Inspect an indexed mesh

Faces contain indices into the vertex list. This compact representation is useful for display and exchange; it does not preserve all of the original analytical geometry.


```python
box = Cell.Prism(width=4, length=3, height=2)
mesh = Topology.MeshData(box)
print("Mesh vertices:", len(mesh["vertices"]))
print("Mesh faces:", len(mesh["faces"]))
print("First face indices:", mesh["faces"][0])
assert len(mesh["faces"]) > 0
for face in mesh["faces"]:
    assert len(face) == 3
```
