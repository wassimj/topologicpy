# 12. IFC and BIM

Read a supplied miniature IFC building with one room. Inspect its semantic model and extract its space geometry.

**Goal:** Complete one small workflow. Run the cells from top to bottom in a fresh kernel.

**Extra dependencies:** None beyond the base environment.

All examples use metres unless stated otherwise.


```python
import os
os.environ["TOPOLOGICPY_CORE_BACKEND"] = "pythonocc"

import topologicpy
from topologicpy.Core import Core

assert topologicpy.__version__ == "0.9.86", "Use TopologicPy 0.9.86 for these tutorials."
assert type(Core.Backend()).__name__ == "PythonOCCBackend", "Restart with the pythonocc backend."
print("TopologicPy", topologicpy.__version__, "—", type(Core.Backend()).__name__)
```

## Read the IFC model

The supplied file was generated for this tutorial. It contains a project, site, building, storey and a single 4 × 3 × 3 m space, so no large external model is required.


```python
from pathlib import Path
from topologicpy.IFC import IFC

path = Path("data") / "one_room.ifc"
model = IFC.ModelByPath(str(path))
summary = IFC.Summary(model)
print(summary)
assert model is not None
```

## Extract the space geometry

IFC distinguishes semantic objects from their geometric representations. Filter the import to spaces, then verify the resulting volume.


```python
from topologicpy.Topology import Topology
from topologicpy.Cell import Cell

spaces = IFC.TopologiesByPath(str(path), includeTypes=["IfcSpace"], topologyType="cell")
print("Spaces:", len(spaces))
room = spaces[0]
print("Type:", Topology.TypeAsString(room))
print("Volume:", Cell.Volume(room))
assert len(spaces) == 1
assert abs(Cell.Volume(room) - 36) < 0.001
```

## Try it yourself
