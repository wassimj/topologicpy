Build a rectangle and a small solid, then translate a second solid and combine them.

**Goal:** Complete one small workflow. Run the cells from top to bottom in a fresh kernel.

**Extra dependencies:** None beyond the base environment.

All examples use metres unless stated otherwise.


```python
import os
os.environ["TOPOLOGICPY_CORE_BACKEND"] = "pythonocc"

import topologicpy
from topologicpy.Core import Core

assert topologicpy.__version__ == "0.9.86", "Use TopologicPy 0.9.86 for these tutorials."
assert type(Core.Backend()).__name__ == "PythonOCCBackend", "Restart with the pythonocc backend."
print("TopologicPy", topologicpy.__version__, "—", type(Core.Backend()).__name__)
```

## From surfaces to solids

A face is a surface; a cell encloses a volume. The default prism placement is centred on its origin.


```python
from topologicpy.Face import Face
from topologicpy.Cell import Cell
from topologicpy.Topology import Topology

floor = Face.Rectangle(width=4, length=3)
room = Cell.Prism(width=4, length=3, height=2)
print("Floor area:", Face.Area(floor))
print("Room volume:", Cell.Volume(room))
assert Face.Area(floor) == 12
assert Cell.Volume(room) == 24
```

## Translate and combine

These two boxes overlap by one metre along X. A union removes the internal overlap. More advanced construction can assemble touching cells into a cell complex.


```python
extension = Topology.Translate(room, x=3)
combined = Topology.Union(room, extension)
print("Result type:", Topology.TypeAsString(combined))
print("Combined volume:", Cell.Volume(combined))
assert Cell.Volume(combined) == 42
```
