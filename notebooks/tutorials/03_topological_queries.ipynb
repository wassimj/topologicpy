## Traverse a topology

A solid contains faces, edges and vertices. Query these objects instead of assuming their internal order.


```python
from topologicpy.Cell import Cell
from topologicpy.Topology import Topology

box = Cell.Prism(width=4, length=3, height=2)
faces = Topology.Faces(box)
edges = Topology.Edges(box)
vertices = Topology.Vertices(box)
print("Faces, edges, vertices:", len(faces), len(edges), len(vertices))
assert [len(faces), len(edges), len(vertices)] == [6, 12, 8]
```

## Ask about incidence

The host topology supplies the context for an adjacency query. Each boundary edge of this closed box belongs to two faces.


```python
edge = edges[0]
touching_faces = Topology.AdjacentTopologies(edge, box, topologyType="Face")
print("Faces incident to the selected edge:", len(touching_faces))
print("Euler characteristic:", len(vertices) - len(edges) + len(faces))
assert len(touching_faces) == 2
assert len(vertices) - len(edges) + len(faces) == 2
```
