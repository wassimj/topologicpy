# 08. Spatial and Geometric Analysis

Measure the distance between two points and the overlap between two rooms.

**Goal:** Complete one small workflow. Run the cells from top to bottom in a fresh kernel.

**Extra dependencies:** None beyond the base environment.

All examples use metres unless stated otherwise.


```python
import os
os.environ["TOPOLOGICPY_CORE_BACKEND"] = "pythonocc"

import topologicpy
from topologicpy.Core import Core

assert topologicpy.__version__ == "0.9.86", "Use TopologicPy 0.9.86 for these tutorials."
assert type(Core.Backend()).__name__ == "PythonOCCBackend", "Restart with the pythonocc backend."
print("TopologicPy", topologicpy.__version__, "—", type(Core.Backend()).__name__)
```

## Measure distance

Coordinates here are in metres. TopologicPy does not infer units from the numbers.


```python
from topologicpy.Vertex import Vertex
from topologicpy.Cell import Cell
from topologicpy.Topology import Topology

entrance = Vertex.ByCoordinates(0, 0, 0)
desk = Vertex.ByCoordinates(3, 4, 0)
distance = Vertex.Distance(entrance, desk)
print("Straight-line distance:", distance, "m")
assert distance == 5
```

## Measure overlap

An intersection returns the shared region. This pair overlaps by 2 × 3 × 2 metres.


```python
room_a = Cell.Prism(width=4, length=3, height=2)
room_b = Topology.Translate(room_a, x=2)
overlap = Topology.Intersect(room_a, room_b)
print("Overlap volume:", Cell.Volume(overlap), "m³")
assert abs(Cell.Volume(overlap) - 12) < 0.0001
```
