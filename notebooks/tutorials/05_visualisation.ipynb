# 05. Visualisation and Communication

Turn a small room into an interactive figure. Rotate it, zoom in, and inspect the edges.

**Goal:** Complete one small workflow. Run the cells from top to bottom in a fresh kernel.

**Extra dependencies:** None beyond the base environment.

All examples use metres unless stated otherwise.


```python
import os
os.environ["TOPOLOGICPY_CORE_BACKEND"] = "pythonocc"

import topologicpy
from topologicpy.Core import Core

assert topologicpy.__version__ == "0.9.86", "Use TopologicPy 0.9.86 for these tutorials."
assert type(Core.Backend()).__name__ == "PythonOCCBackend", "Restart with the pythonocc backend."
print("TopologicPy", topologicpy.__version__, "—", type(Core.Backend()).__name__)
```

## Create display data

Geometry and its display are separate. Plotly converts topology into traces, then puts those traces in a figure.


```python
from topologicpy.Cell import Cell
from topologicpy.Plotly import Plotly

room = Cell.Prism(width=4, length=3, height=2)
data = Plotly.DataByTopology(room, faceColor="skyblue", faceOpacity=0.65)
figure = Plotly.FigureByData(data, width=750, height=450)
assert len(figure.data) > 0
figure.show(renderer="notebook")
```

## Save a portable presentation

The HTML file and notebook display both embed Plotly so they can be opened without an internet connection.


```python
from pathlib import Path

output = Path("results")
output.mkdir(exist_ok=True)
figure.write_html(str(output / "05_room.html"), include_plotlyjs=True)
print("Saved:", output / "05_room.html")
```
